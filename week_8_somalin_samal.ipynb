```python
import re
import logging
# Logging Configuration (Bonus)
logging.basicConfig(level=logging.INFO, format="%(levelname)s: %(message)s")
```

```python
# 🛠️ TOOL 1: Calculator

def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"
print("Output 1:",calculator("2 + 3"))
print("Output 2:",calculator("10 * 5"))
print("Output 3:",calculator("(100 - 20) / 4"))
```

```python
# 🛠️ TOOL 2: Keyword Extractor

def extract_keywords(text: str) -> list:
    """Extract keywords from text."""
    try:
        words = text.split()
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []
print("Output :",extract_keywords("Artificial Intelligence is transforming healthcare through predictive analytics and automation."))
```

## 🤖 Implement Agent Logic Below

👉 Use conditional routing:
- If query contains "calculate" → use calculator
- If query contains "keywords" → use keyword extractor
- Else → general response

```python
# 🤖 AGENT FUNCTION (TO IMPLEMENT)
import re
import logging

def agent(query: str):
    try:
        query_lower = query.lower()

        # Route to Calculator
        if "calculate" in query_lower:
            expression = query_lower.replace("calculate", "").strip()
            result = calculator(expression)

            return {
                "type": "calculation",
                "result": result
            }

        # Route to Keyword Extractor
        elif "keywords" in query_lower:
            text = re.sub(
                r"extract\s+keywords\s+from",
                "",
                query,
                flags=re.IGNORECASE
            ).strip()

            result = extract_keywords(text)

            return {
                "type": "keywords",
                "result": result
            }

        # General Response
        else:
            return {
                "type": "general",
                "result": f"I understand your question: '{query}'. This is a general query."
            }

    except Exception as e:
        logging.error(str(e))
        return {
            "type": "error",
            "result": str(e)
        }
    return {
        "type": "unknown",
        "result": "Not implemented"
    }
```

```python

import re

def agent(query: str):
    query_lower = query.lower()
    try:
        # Calculator Routing
        if "calculate" in query_lower:
            expression = query_lower.replace("calculate", "").strip()
            result = calculator(expression)
            return {
                "type": "calculation",
                "result": result
            }
        # Keyword Routing
        elif "keyword" in query_lower:
            text = re.sub(
                r"extract\s+keywords\s+from",
                "",
                query,
                flags=re.IGNORECASE
            ).strip()
            result = extract_keywords(text)
            return {
                "type": "keywords",
                "result": result
            }
        # General Queries
        else:
            return {
                "type": "general",
                "result": f"You asked: {query}"
            }
    except Exception as e:
        return {
            "type": "error",
            "result": str(e)
        }
```
